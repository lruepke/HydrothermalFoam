## Thermal conductivity of water

```python
import numpy as np 
import matplotlib.pyplot as plt 
import matplotlib as mpl
mpl.rcParams['font.family'] = 'Arial'
mpl.rcParams['font.size'] = 12
mpl.rcParams['mathtext.fontset'] = 'cm'

from iapws import IAPWS97
T=np.linspace(1,790,400)
P=np.linspace(0.5, 100, 400)
PP,TT=np.meshgrid(P,T)
k=np.zeros_like(TT)
for i in range(0,TT.shape[0]):
    for j in range(0,TT.shape[1]):
        steam=IAPWS97(P=PP[i][j], T=TT[i][j]+273.15)
        k[i][j]=steam.k
```

```python
from matplotlib.ticker import MultipleLocator
fig=plt.figure()
ax = plt.gca()
levels=np.linspace(0,0.7,8)
CS=ax.contourf(TT,PP,k,levels=np.linspace(k.min(),k.max(),50),cmap='RdGy_r')
plt.colorbar(CS,format='%.1f',ticks=levels,label='Thermal conductivity (W/m/K)')
ax.set_xlabel('Temperature ($^{\circ}$C)')
ax.set_ylabel('Pressure (MPa)')
ax.xaxis.set_major_locator(MultipleLocator(100))
ax.xaxis.set_minor_locator(MultipleLocator(20))
ax.yaxis.set_major_locator(MultipleLocator(20))
ax.yaxis.set_minor_locator(MultipleLocator(4))
plt.savefig('../images/SinglePhaseDBSFoam/thermalConductivity_water.jpg',dpi=600,bbox_inches='tight')
# plt.savefig('../images/SinglePhaseDBSFoam/thermalConductivity_water.pdf')
plt.show()
```
